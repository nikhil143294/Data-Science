##  Question1 Identify two NLP techniques the chatbot must use to correctly understand and route these customer messages. Explain why each technique is necessary and describe what would happen to chatbot accuracy if either technique were removed

### Two important Natural Language Processing (NLP) techniques that the chatbot must use are:

        1. Intent Recognition (Intent Classification)

What it does:
Intent recognition identifies the user's goal or purpose from their message. It classifies the query into predefined categories such as:

- Order Tracking → "Where is my order?"
- Order Cancellation → "I want to cancel my order."
- Complaint → "My food arrived cold and late."

Why it is necessary:
The chatbot must know what the customer wants before it can provide the correct response or route the request to the appropriate service. For example:

- "Where is my order?" → Route to the order tracking system.
- "Cancel my order." → Route to the cancellation service.
- "My food arrived cold and late." → Route to customer support or the complaint department.

If removed:
Without intent recognition, the chatbot would not understand the user's purpose. It might:

- Give irrelevant answers.
- Route requests to the wrong department.
- Require human intervention more often.

This would significantly reduce chatbot accuracy and customer satisfaction.

        2. Text Preprocessing (Tokenization, Stopword Removal, and Stemming/Lemmatization)

What it does:
Text preprocessing cleans and standardizes the user's message before analysis. Common preprocessing steps include:

- Tokenization: Splits text into individual words.
- Lowercasing: Converts all text to lowercase.
- Stopword Removal: Removes common words like the, is, my, and.
- Stemming/Lemmatization: Converts words to their root form (e.g., arrived → arrive, cancelled → cancel).

Example:

Input:

"My food arrived cold and late."

After preprocessing:

food, arrive, cold, late

Why it is necessary:
Preprocessing removes unnecessary words and keeps only the important information, making it easier for the chatbot to identify the correct intent and keywords.

If removed:
Without preprocessing:

- The model processes many irrelevant words.
- Different forms of the same word are treated as different words (e.g., cancel, cancelled, cancelling).
- Classification accuracy decreases because the chatbot has more noisy and inconsistent data.

| NLP Technique          | Purpose                                                                 | Effect if Removed                                                                                  |
| ---------------------- | ----------------------------------------------------------------------- | -------------------------------------------------------------------------------------------------- |
| **Intent Recognition** | Identifies the user's request (tracking, cancellation, complaint, etc.) | Chatbot cannot determine what the customer wants, leading to incorrect responses and poor routing. |
| **Text Preprocessing** | Cleans and standardizes text for better understanding                   | Noisy and inconsistent input reduces classification accuracy and increases misclassification.      |


## Question2 Describe the text preprocessing pipeline you would apply to this dataset before feeding it to any NLP model. Justify the order in which you would apply each step, and explain specifically why skipping stopword removal could hurt downstream model performance.

Text Preprocessing Pipeline for Food Delivery Reviews

Before feeding the 50,000 customer reviews into an NLP model, the text should be cleaned and standardized. The preprocessing steps should be applied in the following order:

### Step 1: Convert Text to Lowercase

Purpose:
Convert all letters to lowercase so that words with different cases are treated as the same word.

Example:

- Before: "The FOOD Was Great!"
- After: "the food was great!"

Why first?
Lowercasing ensures consistency before any further processing.

### Step 2: Remove Special Characters, Punctuation, Emojis, and Numbers

Purpose:
Remove unnecessary symbols that usually do not contribute to understanding the review.

Example:

- Before:
"Food was awesome!!! 😍 Delivered in 25 mins."
- After:
"food was awesome delivered in mins"

Why second?
Cleaning the text removes noise before splitting it into words.

### Step 3: Tokenization

Purpose:
Split the cleaned text into individual words (tokens).

Example:

- Input:
"food was delivered late"
- Output:
["food", "was", "delivered", "late"]

Why here?
Most later preprocessing steps operate on individual words rather than whole sentences.

### Step 4: Stopword Removal

Purpose:
Remove common words such as the, is, a, an, and, was, of, to, which usually carry little meaning.

Example:

- Before:
["the", "food", "was", "very", "good"]
- After:
["food", "good"]

Why after tokenization?
Stopwords are identified as individual tokens, so tokenization must occur first.

### Step 5: Stemming or Lemmatization

Purpose:
Reduce words to their root or base form.

Examples:

- Stemming:
delivered → deliver
delivering → deliver
- Lemmatization:
better → good
running → run

Why after stopword removal?
Only meaningful words are processed, reducing unnecessary computation.

### Final Pipeline

1.  Convert text to lowercase
2. Remove special characters, punctuation, emojis, and numbers
3. Tokenize the text
4. Remove stopwords
5. Apply stemming or lemmatization

### Why This Order?

Each step prepares the text for the next:

- Lowercasing standardizes the text.
- Cleaning removes unwanted symbols.
- Tokenization converts sentences into words.
- Stopword removal eliminates unimportant words.
- Stemming/Lemmatization reduces similar words to a common form.

Applying the steps in this order produces clean, consistent, and meaningful input for the NLP model.

### Why Skipping Stopword Removal Hurts Model Performance

If stopword removal is skipped:

- The dataset contains many high-frequency but low-information words such as the, is, a, and, was.
- These words increase the vocabulary size unnecessarily.
- Important keywords like late, cold, refund, cancel, and delivery become less prominent.
- The model spends time learning patterns from meaningless words instead of useful ones.
- Training becomes slower and requires more memory.
- Classification accuracy may decrease because the model has more noise and fewer informative features.

### Example:

Original review:

"The food was cold and the delivery was late."

Without stopword removal:

["the", "food", "was", "cold", "and", "the", "delivery", "was", "late"]

With stopword removal:

["food", "cold", "delivery", "late"]

## Question3 Compare Bag of Words and TF-IDF as text representation methods for this ranking task. Which would you choose and why? Identify one limitation that your chosen method still has even after it is applied.

### Comparison of Bag of Words (BoW) and TF-IDF for Food Item Search

When a customer searches for "spicy chicken without onion", the search system must represent both the query and menu descriptions in a numerical form so that they can be compared.

        1. Bag of Words (BoW)

What it is:
Bag of Words represents text by counting how many times each word appears in a document, while ignoring grammar and word order.

Example:

Query:

"spicy chicken without onion"

BoW representation:

- spicy = 1
- chicken = 1
- without = 1
- onion = 1

Advantages:

- Simple to implement.
- Fast to compute.
- Works well for basic text classification tasks.

Disadvantages:

- Treats all words as equally important.
- Very common words may dominate the representation.
- Does not consider how informative a word is across all menu descriptions.

        2. TF-IDF (Term Frequency–Inverse Document Frequency)

What it is:
TF-IDF assigns a weight to each word based on:

- Term Frequency (TF): How often the word appears in a document.
- Inverse Document Frequency (IDF): How rare the word is across all documents.

Words that are common across many menu descriptions receive lower weights, while unique and informative words receive higher weights.

Example:

Query:

"spicy chicken without onion"

Important words like spicy and onion receive higher weights than common words.

Advantages:

- Highlights important keywords.
- Reduces the influence of common words.
- Produces better relevance ranking than BoW.
- Widely used in search engines and information retrieval.

Disadvantages:

- Slightly more computationally expensive than BoW.
- Still ignores the meaning and order of words.

| Feature                                    | Bag of Words (BoW) | TF-IDF                   |
| ------------------------------------------ | ------------------ | ------------------------ |
| Representation                             | Word counts        | Weighted word importance |
| Considers word frequency                   | Yes                | Yes                      |
| Considers word importance across documents | No                 | Yes                      |
| Handles common words well                  | No                 | Yes                      |
| Search ranking quality                     | Basic              | Better                   |
| Computational cost                         | Low                | Moderate                 |


### Which Method Would I Choose?

I would choose TF-IDF for the food item search feature.

Reason:
TF-IDF gives higher importance to meaningful keywords such as spicy, chicken, and onion, while reducing the weight of very common words. This helps rank the most relevant menu items higher and provides more accurate search results than Bag of Words.

For example, when searching for:

"spicy chicken without onion"

TF-IDF is more likely to rank menu descriptions containing spicy and chicken higher than descriptions containing only common words.

### One Limitation of TF-IDF

Even after applying TF-IDF, it does not understand the meaning or context of words.

For example:

Query: "spicy chicken without onion"
Menu: "Spicy chicken with onion"

Both descriptions contain nearly the same words, so TF-IDF may assign them similar relevance scores. However, it cannot understand that "without onion" and "with onion" have opposite meanings.

## Question4 Explain the core probabilistic assumption that Naive Bayes makes about its input features. Describe a realistic food delivery example where this assumption is violated, and explain how that violation would affect the model's classification accuracy.

### Naive Bayes Assumption

The core probabilistic assumption made by the Naive Bayes classifier is the conditional independence assumption.

This means that all input features (words) are assumed to be independent of each other given the class label. In other words, once the class (e.g., Delivery Issue, Food Quality Issue, or App Issue) is known, the occurrence of one word does not affect the occurrence of another word.

For example, in a ticket classified as Food Quality Issue, Naive Bayes assumes that the words "cold", "food", and "late" occur independently, even though they may actually be related.

### Realistic Food Delivery Example

Consider the customer support ticket:

"My food arrived cold and late."

This ticket belongs to the Food Quality Issue category.

The important words are:

- food
- arrived
- cold
- late

In reality, the words "cold" and "late" are highly related because food often becomes cold due to a late delivery. Their occurrences are not independent.

However, Naive Bayes treats them as completely independent features when calculating probabilities.

### How the Assumption Is Violated

The independence assumption is violated because:

- A late delivery increases the likelihood that the food is cold.
- The words "late" and "cold" frequently appear together in customer complaints.
- Their occurrences are correlated rather than independent.

Thus, the real-world relationship between these words is ignored by Naive Bayes.

### Effect on Classification Accuracy

When the independence assumption is violated:

- The model may incorrectly estimate the probability of each class.
- Correlated words may be treated as separate pieces of evidence, causing the model to overestimate or underestimate the importance of those words.
- Similar complaints may be assigned to the wrong category.

For example:

"Food arrived cold and late."

This ticket could be classified as either:

- Food Quality Issue (because of cold), or
- Delivery Issue (because of late).

Since Naive Bayes ignores the relationship between cold and late, it may not correctly determine which category is most appropriate, reducing classification accuracy.




## Question5 Describe two evaluation metrics you would prioritise over accuracy when comparing these models on this imbalanced dataset. Explain one technique you would apply to address the class imbalance before training.

### Evaluation Metrics for an Imbalanced Dataset

The dataset contains:

- Positive: 80%
- Neutral: 15%
- Negative: 5%

Since the dataset is highly imbalanced, accuracy is not a reliable evaluation metric. For example, a model that predicts every review as Positive would achieve 80% accuracy, even though it completely fails to identify Neutral and Negative reviews.

Instead, I would prioritise the following metrics:

        1. Precision

Definition:
Precision measures the proportion of reviews predicted as a particular class that are actually correct.

Formula:

Precision=TP+FP/TP
	​


Where:

- TP = True Positives
- FP = False Positives

Why it is important:
Precision is especially useful for the Negative class. If the model labels a review as negative, we want to be confident that it is truly negative. High precision reduces false alarms.

        2. F1-Score

Definition:
F1-score is the harmonic mean of Precision and Recall, providing a balanced measure of model performance.

	​


Why it is important:
F1-score is well suited to imbalanced datasets because it considers both:

- Precision (avoiding false positives)
- Recall (detecting as many true cases as possible)

A macro-averaged F1-score evaluates performance across all classes equally, including the minority Negative class.

### Technique to Handle Class Imbalance

SMOTE (Synthetic Minority Over-sampling Technique)

What it does:
SMOTE increases the number of samples in minority classes by generating synthetic examples rather than simply duplicating existing ones.

For this dataset:

Before balancing:

- Positive = 80%
- Neutral = 15%
- Negative = 5%

After applying SMOTE:

- More synthetic Neutral and Negative reviews are created, making the class distribution more balanced.

Benefits:

- Gives the model more examples of minority classes.
- Improves the ability to recognise Neutral and Negative reviews.
- Reduces bias toward predicting the majority Positive class.


## Question6 Explain what the attention mechanism in transformers enables that traditional RNN-based models could not do effectively with long text sequences. Describe one practical challenge of running a transformer model in a high-traffic production environment and one common way to mitigate it.

### Attention Mechanism in Transformers

The attention mechanism allows a transformer model to focus on the most relevant words in a sentence when processing text. Instead of reading words one by one like an RNN, the transformer can consider all words in the sequence simultaneously and determine which ones are most important for understanding the context.

  For example, consider the review:

"The pizza was delicious, but it arrived 45 minutes late, so I want a refund."

The attention mechanism can directly relate the words "refund" and "late", even though they are far apart in the sentence. This helps the model generate an appropriate response, such as apologising for the delayed delivery and offering assistance.

### Why Transformers Perform Better Than RNNs

Traditional RNN-based models process text sequentially, one word at a time.

This creates two major problems with long text:

- Difficulty remembering long-range context: Information from earlier words may fade as the sequence becomes longer (vanishing gradient problem).
- Slow processing: Since words are processed one after another, computation cannot be fully parallelised.

The attention mechanism overcomes these limitations by:

- Looking at all words in the sequence at the same time.
- Capturing relationships between distant words.
- Preserving long-range context more effectively.
- Allowing parallel processing, making training and inference faster.

As a result, transformers generally achieve better performance on long text sequences than RNNs.

### Practical Challenge in a High-Traffic Production Environment

A major challenge is high computational cost and latency.

In a food delivery app handling hundreds of customer requests per minute, transformer models:

- Require significant CPU/GPU resources.
- Consume more memory than traditional models.
- Can increase response time if many requests arrive simultaneously.

This may lead to slower customer responses and higher infrastructure costs.

### One Common Mitigation Technique

A common solution is using a smaller or optimized transformer model.

Examples include:

- DistilBERT
- MobileBERT
- Quantized or pruned transformer models

These models:

- Reduce memory usage.
- Perform inference faster.
- Lower hardware costs.
- Maintain accuracy close to that of larger transformer models.

This enables the application to handle high request volumes while still generating high-quality responses in real time.
